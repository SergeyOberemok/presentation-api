```python
import json

from faker import Faker
```

```python
fake = Faker()
types = ['text']
```

```python
def create_slide_item(index: int) -> dict[str, str]:
    return {'type': types[0], 'value': f'{str(index)} - {fake.sentence()}'}
```

```python
def create_text_items_list(quantity: int) -> list[dict[str, str]]:
    return [create_slide_item(i) for i in range(quantity)]
```

```python
quantity = 0
slides_list = []

try:
    quantity = int(input('How many items do you want? ')) or 0

    slides_list = create_text_items_list(quantity)
except ValueError:
    print('Enter a valid quantity')
```

```python
# json.dumps(slides_list)
```

```python
with open('slides.json', 'w') as outfile:
    json.dump(slides_list, outfile)
```
